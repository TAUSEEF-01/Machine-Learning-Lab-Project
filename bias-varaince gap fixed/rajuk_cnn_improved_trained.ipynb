# Rajuk Image Level Classifier — Improved CNN (Bias–Variance Reduced)

**Improvements over previous version:**
1. ✅ **ResNet18 Transfer Learning** — pretrained ImageNet weights
2. ✅ **On-the-fly augmentation** via `torchvision.transforms` — diverse, applied per-epoch
3. ✅ **Weighted CrossEntropyLoss + label smoothing** — handles class imbalance, softens over-confident targets
4. ✅ **AdamW + weight decay** — stronger regularisation against overfitting
5. ✅ **Cosine LR scheduler** — smoother learning rate decay
6. ✅ **Leakage-free split** — augmented images stay strictly in training set

**New bias–variance reduction techniques (this version):**
7. ✅ **Bagging (bootstrap aggregating)** — `N_ENSEMBLE` models are each trained on a bootstrap resample of the training set and their softmax outputs are averaged (soft voting). Averaging models that make *independent* errors directly cancels **variance**.
8. ✅ **Out-of-bag (OOB) early stopping** — each member early-stops on the ~37 % of training images its bootstrap sample missed, so the validation set is never used while building the ensemble.
9. ✅ **K-fold model recycling** — the cross-validation models from section 11 are added to the ensemble for free extra diversity.
10. ✅ **MixUp regularisation** — trains on convex blends of image pairs → smoother decision boundaries, lower **variance**.
11. ✅ **Flip test-time augmentation (TTA)** — predictions averaged over the image and its mirror → less prediction noise.
12. ✅ **layer3 unfrozen + discriminative LRs** — more trainable capacity (lower **bias**) at a reduced backbone LR so the extra capacity does not blow up the variance.

## 4. Load Labels from Excel / CSV

```python
def normalize_header(name):
    return ''.join(ch for ch in str(name).lower() if ch.isalnum())

def find_column(df, target_normalized):
    for col in df.columns:
        if normalize_header(col) == target_normalized:
            return col
    return None

def normalize_level(value):
    if pd.isna(value):
        return None
    text = str(value).strip().lower()
    mapping = {'p': 'positive', 'positive': 'positive',
               'm': 'moderate',  'moderate': 'moderate',
               'n': 'negative',  'negative': 'negative'}
    return mapping.get(text)

def load_label_mapping(df, level_col, class_names):
    label_lookup = {name: idx for idx, name in enumerate(class_names)}
    label_map = {}
    for row_index, value in enumerate(df[level_col].tolist(), start=1):
        normalized = normalize_level(value)
        if normalized is None:
            continue
        label_map[row_index] = label_lookup[normalized]
    return label_map

if XLSX_PATH.exists():
    df = pd.read_excel(XLSX_PATH)
    source_path = XLSX_PATH
elif CSV_PATH.exists():
    df = pd.read_csv(CSV_PATH)
    source_path = CSV_PATH
else:
    raise FileNotFoundError(
        f'No dataset file found at {DRIVE_BASE}. '
        'Expected Rajuk_ML_DataSet.xlsx or Real ML_DataSet - Sheet1.csv'
    )

level_col = (
    find_column(df, 'levelpositivemoderatenegative')
    or find_column(df, 'level')
)
if level_col is None:
    raise ValueError('Required level column not found in the dataset.')

label_map = load_label_mapping(df, level_col, CLASS_NAMES)
print(f'Loaded {len(df):,} rows  |  {len(label_map):,} valid labels  |  source: {source_path.name}')
```

## 5. Collect Image Paths & Leakage-Free Split

Split is done by **unique subject ID** (real images only) so that no subject's augmented
copy can appear in val/test. Augmented images are assigned to training only.

```python
def parse_image_index(stem):
    match = re.match(r'img(\d+)', stem.lower())
    return int(match.group(1)) if match else None

def collect_all_raw(base_dir, label_map):
    paths, labels = [], []
    skip_idx = skip_lbl = 0
    if not base_dir.exists():
        return paths, labels, skip_idx, skip_lbl
    for p in base_dir.rglob('*'):
        if not (p.is_file() and p.suffix.lower() in ALLOWED_EXTS):
            continue
        idx = parse_image_index(p.stem.strip())
        if idx is None:
            skip_idx += 1; continue
        lbl = label_map.get(idx)
        if lbl is None:
            skip_lbl += 1; continue
        paths.append(p)
        labels.append(lbl)
    return paths, labels, skip_idx, skip_lbl

aug_paths,  aug_labels,  si_a, sl_a = collect_all_raw(AUGMENTED_DIR, label_map)
real_paths, real_labels, si_r, sl_r = collect_all_raw(REAL_DIR,      label_map)

print(f'Augmented images         : {len(aug_paths):,}')
print(f'Real images              : {len(real_paths):,}')
print(f'Skipped (no index)       : {si_a + si_r}')
print(f'Skipped (no label match) : {sl_a + sl_r}')

# ── Leakage-free split by unique subject ID ────────────────────────────────────
unique_indices     = sorted(set(
    parse_image_index(p.stem.strip()) for p in real_paths
    if parse_image_index(p.stem.strip()) is not None
))
unique_labels_list = [label_map[i] for i in unique_indices]
all_uq             = np.arange(len(unique_indices))

trainval_uq, test_uq = train_test_split(
    all_uq, test_size=TEST_FRACTION, random_state=SEED, stratify=unique_labels_list
)
val_rel = VAL_FRACTION / (1.0 - TEST_FRACTION)
train_uq, val_uq = train_test_split(
    trainval_uq, test_size=val_rel, random_state=SEED,
    stratify=np.array(unique_labels_list)[trainval_uq]
)

train_img_ids = set(np.array(unique_indices)[train_uq])
val_img_ids   = set(np.array(unique_indices)[val_uq])
test_img_ids  = set(np.array(unique_indices)[test_uq])

train_paths_final, train_labels_final = [], []
val_paths_final,   val_labels_final   = [], []
test_paths_final,  test_labels_final  = [], []

# Augmented → each subject's aug copies follow it into its split
for p, lbl in zip(aug_paths, aug_labels):
    img_id = parse_image_index(p.stem.strip())
    if img_id in train_img_ids:
        train_paths_final.append(p); train_labels_final.append(lbl)
    elif img_id in val_img_ids:
        val_paths_final.append(p);   val_labels_final.append(lbl)
    elif img_id in test_img_ids:
        test_paths_final.append(p);  test_labels_final.append(lbl)

# Real → correct split
for p, lbl in zip(real_paths, real_labels):
    img_id = parse_image_index(p.stem.strip())
    if img_id in train_img_ids:
        train_paths_final.append(p); train_labels_final.append(lbl)
    elif img_id in val_img_ids:
        val_paths_final.append(p);   val_labels_final.append(lbl)
    elif img_id in test_img_ids:
        test_paths_final.append(p);  test_labels_final.append(lbl)

N_total = len(train_paths_final) + len(val_paths_final) + len(test_paths_final)
print(f'\nLeakage-free split:')
print(f'  Train (real + aug) : {len(train_paths_final):,}  ({100*len(train_paths_final)/N_total:.1f} %)')
print(f'  Val   (real + aug) : {len(val_paths_final):,}  ({100*len(val_paths_final)/N_total:.1f} %)')
print(f'  Test  (real + aug) : {len(test_paths_final):,}  ({100*len(test_paths_final)/N_total:.1f} %)')
print(f'  Total accounted    : {N_total:,}  (of {len(aug_paths)+len(real_paths):,} raw — diff = {len(aug_paths)+len(real_paths)-N_total} skipped)')

val_counts  = np.bincount(val_labels_final,  minlength=len(CLASS_NAMES))
test_counts = np.bincount(test_labels_final, minlength=len(CLASS_NAMES))
print('\nValidation-set class distribution:')
for name, cnt in zip(CLASS_NAMES, val_counts):
    print(f'  {name:10s}: {cnt:,}')
print('\nTest-set class distribution:')
for name, cnt in zip(CLASS_NAMES, test_counts):
    print(f'  {name:10s}: {cnt:,}')

all_labels_combined = train_labels_final + val_labels_final + test_labels_final
counts = np.bincount(all_labels_combined, minlength=len(CLASS_NAMES))
print('\nClass distribution (total):')
for name, cnt in zip(CLASS_NAMES, counts):
    print(f'  {name:10s}: {cnt:,}')
```

## 5b. Hard-Example Mining — Inject Misclassified Validation Images into TRAIN

The previous run exported every image the model got wrong on the validation set to
`val_misclassified.csv`. Here we feed those *hard examples* back into the **training
set** so the network gets a second chance to learn them:

* **all** misclassified **raw / real** images  → moved into TRAIN
* **some** of their **augmented** copies (a per-subject sample controlled by
  `AUG_KEEP_FRACTION`) → moved into TRAIN

Matched images are pulled out of the val / test lists and appended to the train lists
*before* the datasets are cached, so the change flows through the entire pipeline
(weighted loss, k-fold, learning curves, final report).

> **Note:** because only *some* augmented copies of a hard subject are moved while the
> rest stay in validation, those specific hard subjects can appear in both splits. This
> is a deliberate trade-off of hard-example mining (the goal is to *learn the hard
> cases*), not a clean random-split guarantee.

```python
# ── Hard-example mining: move misclassified validation images into TRAIN ───
from collections import defaultdict

MISCLASSIFIED_CSV = OUTPUT_DIR / 'val_misclassified.csv'
AUG_KEEP_FRACTION = 0.5     # fraction of each subject's misclassified AUG copies -> train

def _norm_path(p):
    return os.path.normcase(os.path.normpath(str(p)))

if not MISCLASSIFIED_CSV.exists():
    print(f'⚠  {MISCLASSIFIED_CSV.name} not found — skipping hard-example mining.')
else:
    hard_df = pd.read_csv(MISCLASSIFIED_CSV)
    print(f'Loaded {len(hard_df):,} misclassified rows from {MISCLASSIFIED_CSV.name}')

    # Split the misclassified rows into raw (real) vs augmented
    real_mask = hard_df['image_type'].astype(str).str.lower().eq('real')
    hard_real_paths = {_norm_path(p) for p in hard_df.loc[real_mask, 'file_path']}

    # Group augmented misclassified paths by subject id, then sample "some" per subject
    aug_by_subject = defaultdict(list)
    for _, r in hard_df.loc[~real_mask].iterrows():
        aug_by_subject[r['image_id']].append(_norm_path(r['file_path']))

    rng = np.random.RandomState(SEED)
    hard_aug_selected = set()
    for iid, plist in aug_by_subject.items():
        k = max(1, int(round(len(plist) * AUG_KEEP_FRACTION)))
        chosen = rng.choice(len(plist), size=min(k, len(plist)), replace=False)
        hard_aug_selected.update(plist[j] for j in chosen)

    move_set = hard_real_paths | hard_aug_selected

    # Pull matched paths out of val / test; collect them to append to train
    def _extract(paths, labels):
        keep_p, keep_l, moved = [], [], []
        for p, l in zip(paths, labels):
            if _norm_path(p) in move_set:
                moved.append((p, l))
            else:
                keep_p.append(p); keep_l.append(l)
        return keep_p, keep_l, moved

    val_paths_final,  val_labels_final,  moved_val  = _extract(val_paths_final,  val_labels_final)
    test_paths_final, test_labels_final, moved_test = _extract(test_paths_final, test_labels_final)

    already_in_train = {_norm_path(p) for p in train_paths_final}
    n_already = len(move_set & already_in_train)

    for p, l in moved_val + moved_test:
        train_paths_final.append(p); train_labels_final.append(l)

    n_aug_total = sum(len(v) for v in aug_by_subject.values())
    print(f'  Misclassified RAW images         : {len(hard_real_paths):,}  (all → train)')
    print(f'  Misclassified AUG images (total) : {n_aug_total:,}')
    print(f'  AUG copies selected for train    : {len(hard_aug_selected):,}  '
          f'(≈{AUG_KEEP_FRACTION:.0%} per subject)')
    print(f'  Moved into TRAIN  (from val)     : {len(moved_val):,}')
    print(f'  Moved into TRAIN  (from test)    : {len(moved_test):,}')
    print(f'  Already in TRAIN (split changed) : {n_already:,}')

    N_total = len(train_paths_final) + len(val_paths_final) + len(test_paths_final)
    print('\nSplit after hard-example mining:')
    print(f'  Train : {len(train_paths_final):,}  ({100*len(train_paths_final)/N_total:.1f} %)')
    print(f'  Val   : {len(val_paths_final):,}  ({100*len(val_paths_final)/N_total:.1f} %)')
    print(f'  Test  : {len(test_paths_final):,}  ({100*len(test_paths_final)/N_total:.1f} %)')
```

## 7. Dataset with On-the-fly Augmentation

**Why on-the-fly instead of pre-generated files?**
- Every epoch the model sees a *differently augmented* version of each image → more effective regularisation
- No disk bloat from 16x augmented copies
- Val/test use only resize + normalise (no augmentation)

```python
# ── Transforms ────────────────────────────────────────────────────────────────
train_transform = T.Compose([
    T.Resize(IMAGE_SIZE),
    T.RandomHorizontalFlip(p=0.5),
    T.RandomVerticalFlip(p=0.3),
    T.RandomRotation(degrees=15),
    T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.1),
    T.RandomResizedCrop(IMAGE_SIZE, scale=(0.75, 1.0)),
    T.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0)),
    T.RandomGrayscale(p=0.05),
    T.ToTensor(),
    T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

val_test_transform = T.Compose([
    T.Resize(IMAGE_SIZE),
    T.ToTensor(),
    T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])


class RajukDataset(Dataset):
    """
    Caches images as PIL uint8 objects (NOT float32 tensors).
    RAM cost: N * H * W * 3 * 1 byte  (12x cheaper than float32 tensors)

    For 2700 images at 224x224:
      PIL uint8   → 0.41 GB  ✅ safe on 12.7 GB RAM
      float32     → 1.63 GB  ❌ what caused the crash before

    Transforms applied fresh each __getitem__ → different augmentation every epoch.
    """

    def __init__(self, image_paths, labels, transform):
        self.labels    = labels
        self.transform = transform
        n = len(image_paths)
        ram_mb = n * IMAGE_SIZE[0] * IMAGE_SIZE[1] * 3 / 1e6
        print(f'  Caching {n:,} PIL images at {IMAGE_SIZE} … (est. {ram_mb:.0f} MB)')
        self.images = []
        for p in image_paths:
            with Image.open(p) as img:
                # Store as resized uint8 PIL — smallest possible in-memory form
                self.images.append(img.convert('RGB').resize(IMAGE_SIZE, Image.BILINEAR))
        print(f'  Done. All {n:,} images in RAM.')

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        # Fresh transform each call = on-the-fly augmentation every epoch
        return self.transform(self.images[idx]), self.labels[idx]


if len(train_paths_final) == 0:
    raise ValueError('No labeled images found. Check folder paths and label mapping.')

print('Building datasets (PIL uint8 RAM cache at 224x224):')
train_dataset = RajukDataset(train_paths_final, train_labels_final, train_transform)
val_dataset   = RajukDataset(val_paths_final,   val_labels_final,   val_test_transform)
test_dataset  = RajukDataset(test_paths_final,  test_labels_final,  val_test_transform)

total_mb = (len(train_dataset)+len(val_dataset)+len(test_dataset)) * 224*224*3 / 1e6
print(f'\nDatasets ready — Train: {len(train_dataset):,}  '
      f'Val: {len(val_dataset):,}  Test: {len(test_dataset):,}')
print(f'Total image RAM: ~{total_mb:.0f} MB  (out of 12,700 MB available)')
```

## 9. ResNet18 Transfer Learning Model

With only ~160 real images, training from scratch always overfits.
ResNet18 pretrained on ImageNet already knows edges, textures, and shapes —
we only need to teach it to distinguish our 3 building-condition classes.

**Strategy:** Freeze all layers except the last residual block (`layer4`) and the
classifier head. This gives fast convergence and strong regularisation.

```python
class RajukResNet(nn.Module):
    """
    ResNet18 backbone with frozen early layers.
    layer4 (+ layer3 when UNFREEZE_LAYER3=True) and the classifier head train.

    Unfreezing layer3 adds capacity where the pretrained features are still
    fairly generic → lower BIAS. The bagging ensemble, MixUp and label
    smoothing keep the extra VARIANCE in check.
    """

    def __init__(self, num_classes: int, dropout: float = 0.5):
        super().__init__()
        backbone = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)

        trainable_prefixes = ['layer4', 'fc'] + (['layer3'] if UNFREEZE_LAYER3 else [])
        for name, param in backbone.named_parameters():
            if not any(name.startswith(x) for x in trainable_prefixes):
                param.requires_grad = False

        # Replace final FC with dropout + linear
        in_features  = backbone.fc.in_features   # 512 for ResNet18
        backbone.fc  = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(in_features, num_classes)
        )
        self.model = backbone

    def forward(self, x):
        return self.model(x)


_tmp = RajukResNet(len(CLASS_NAMES))
total_params     = sum(p.numel() for p in _tmp.parameters())
trainable_params = sum(p.numel() for p in _tmp.parameters() if p.requires_grad)
print(f'Total parameters     : {total_params:,}')
print(f'Trainable parameters : {trainable_params:,}  '
      f'({100*trainable_params/total_params:.1f} % of total)')
del _tmp
```

## 10. Weighted Loss & Training Helpers

Class weights are inversely proportional to class frequency so the model
doesn't just predict 'negative' (the majority class) for everything.

```python
# ── Weighted CrossEntropyLoss (+ label smoothing → variance reduction) ───────
train_labels_arr = np.array(train_labels_final, dtype=np.int64)
class_counts     = np.bincount(train_labels_arr, minlength=len(CLASS_NAMES))
class_weights    = torch.tensor(
    1.0 / (class_counts + 1e-6), dtype=torch.float32
).to(device)
class_weights   /= class_weights.sum()   # normalise to sum=1

criterion = nn.CrossEntropyLoss(weight=class_weights, label_smoothing=LABEL_SMOOTHING)

print('Class counts in training set:')
for name, cnt, w in zip(CLASS_NAMES, class_counts, class_weights.cpu()):
    print(f'  {name:10s}: {cnt:,} samples  →  weight {w:.4f}')


# ── MixUp (variance reduction) ────────────────────────────────────────────────
def mixup_batch(X, y, alpha):
    """Convex blend of the batch with a shuffled copy of itself."""
    lam  = float(np.random.beta(alpha, alpha))
    perm = torch.randperm(X.size(0), device=X.device)
    return lam * X + (1.0 - lam) * X[perm], y, y[perm], lam


# ── Train / evaluate functions ────────────────────────────────────────────────
def train_one_epoch(model, loader, optimizer, use_mixup=True):
    model.train()
    total_loss = correct = total = 0
    for X, y in loader:
        X, y = X.to(device), y.to(device)
        optimizer.zero_grad()
        if use_mixup and MIXUP_ALPHA > 0 and np.random.rand() < MIXUP_PROB:
            Xm, y_a, y_b, lam = mixup_batch(X, y, MIXUP_ALPHA)
            logits = model(Xm)
            loss   = lam * criterion(logits, y_a) + (1.0 - lam) * criterion(logits, y_b)
            preds  = logits.argmax(1)
            correct += lam * (preds == y_a).sum().item() \
                     + (1.0 - lam) * (preds == y_b).sum().item()
        else:
            logits = model(X)
            loss   = criterion(logits, y)
            correct += (logits.argmax(1) == y).sum().item()
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * X.size(0)
        total      += X.size(0)
    return total_loss / total, correct / total


@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    total_loss = correct = total = 0
    all_preds, all_true = [], []
    for X, y in loader:
        X, y = X.to(device), y.to(device)
        logits = model(X)
        loss   = criterion(logits, y)
        total_loss += loss.item() * X.size(0)
        preds = logits.argmax(1)
        correct += (preds == y).sum().item()
        total   += X.size(0)
        all_preds.extend(preds.cpu().numpy())
        all_true.extend(y.cpu().numpy())
    return total_loss / total, correct / total, all_preds, all_true


def make_model():
    """AdamW with discriminative LRs (backbone slower than head) + cosine schedule."""
    m = RajukResNet(len(CLASS_NAMES), dropout=DROPOUT).to(device)
    backbone_params = [p for n, p in m.named_parameters()
                       if p.requires_grad and not n.startswith('model.fc')]
    head_params     = [p for n, p in m.named_parameters()
                       if p.requires_grad and n.startswith('model.fc')]
    opt = optim.AdamW(
        [{'params': backbone_params, 'lr': LR * LR_BACKBONE_MULT},
         {'params': head_params,     'lr': LR}],
        weight_decay=WEIGHT_DECAY
    )
    sch = optim.lr_scheduler.CosineAnnealingLR(opt, T_max=NUM_EPOCHS, eta_min=1e-6)
    return m, opt, sch


print('\nHelpers defined (label smoothing + MixUp + discriminative LRs active).')
```

## 11. K-Fold Cross-Validation on Train Split

```python
min_cls = np.min(np.bincount(train_labels_arr, minlength=len(CLASS_NAMES)))
k       = min(KFOLD_SPLITS, int(min_cls))
kfold   = StratifiedKFold(n_splits=k, shuffle=True, random_state=SEED)

train_indices_all = np.arange(len(train_dataset))

fold_val_accs     = []
fold_states       = []
best_global_acc   = -1.0
best_global_state = None

for fold, (tr, va) in enumerate(kfold.split(train_indices_all, train_labels_arr)):
    print(f'\n{"─"*58}')
    print(f'  Fold {fold+1}/{k}   |   train={len(tr):,}   val={len(va):,}')
    print(f'{"─"*58}')

    fold_train_loader = DataLoader(
        Subset(train_dataset, tr),
        batch_size=BATCH_SIZE, shuffle=True, num_workers=NW, pin_memory=pin
    )
    fold_val_loader = DataLoader(
        Subset(train_dataset, va),
        batch_size=BATCH_SIZE, shuffle=False, num_workers=NW, pin_memory=pin
    )

    model, optimizer, scheduler = make_model()
    best_fold_acc   = -1.0
    best_fold_state = None
    patience_ctr    = 0

    for epoch in range(NUM_EPOCHS):
        tr_loss, tr_acc = train_one_epoch(model, fold_train_loader, optimizer)
        vl_loss, vl_acc, _, _ = evaluate(model, fold_val_loader)
        scheduler.step()

        print(f'  Ep {epoch+1:02d}/{NUM_EPOCHS}  '
              f'Train Loss={tr_loss:.4f} Acc={tr_acc:.4f}  |  '
              f'Val   Loss={vl_loss:.4f} Acc={vl_acc:.4f}')

        if vl_acc > best_fold_acc:
            best_fold_acc   = vl_acc
            patience_ctr    = 0
            best_fold_state = {k2: v.cpu() for k2, v in model.state_dict().items()}
        else:
            patience_ctr += 1
            if patience_ctr >= PATIENCE:
                print(f'  Early stopping at epoch {epoch+1}.')
                break

    fold_val_accs.append(best_fold_acc)
    fold_states.append(best_fold_state)
    print(f'  → Fold {fold+1} best val acc: {best_fold_acc:.4f}')

    if best_fold_acc > best_global_acc:
        best_global_acc   = best_fold_acc
        best_global_state = best_fold_state

print(f'\nCross-val accuracies : {[f"{a:.4f}" for a in fold_val_accs]}')
print(f'Mean CV accuracy     : {np.mean(fold_val_accs):.4f} ± {np.std(fold_val_accs):.4f}')
```

## 12. Bagging — Bootstrap-Aggregated Ensemble → Evaluate on Val + Test

A single CNN trained on ~1k images has high **variance**: retrain it on a slightly
different sample and its predictions change noticeably. **Bagging (bootstrap
aggregating)** attacks exactly that:

1. Draw `N_ENSEMBLE` **bootstrap samples** of the training set — same size, drawn
   *with replacement*, stratified per class.
2. Train one model per sample. Each model early-stops on its **out-of-bag (OOB)**
   images (the ~37 % its bootstrap missed) — the real validation set is never used
   to build the ensemble.
3. **Soft voting**: average the members' softmax probabilities. Errors that are
   independent across members cancel out, so the variance part of the error drops
   roughly like 1/N.
4. The k-fold models from section 11 are added as extra members, and predictions
   are averaged over horizontal-flip views (**TTA**) for a further variance cut.


```python
# ── Bagging: train N_ENSEMBLE models on bootstrap resamples ──────────────────
class _DetView(Dataset):
    """Deterministic (non-augmenting) view of the cached train dataset,
    used to score each member on its out-of-bag (OOB) images."""
    def __init__(self, base_ds, indices):
        self.base, self.indices = base_ds, list(indices)
    def __len__(self):
        return len(self.indices)
    def __getitem__(self, i):
        j = self.indices[i]
        return val_test_transform(self.base.images[j]), self.base.labels[j]


rng_boot         = np.random.RandomState(SEED)
bootstrap_states = []
oob_accs         = []
all_idx          = np.arange(len(train_dataset))

for m_idx in range(N_ENSEMBLE):
    # Stratified bootstrap: sample each class WITH replacement → same class mix
    boot_parts = []
    for c in range(len(CLASS_NAMES)):
        cls_idx = np.where(train_labels_arr == c)[0]
        n_draw  = max(1, int(round(len(cls_idx) * BOOTSTRAP_FRAC)))
        boot_parts.append(rng_boot.choice(cls_idx, size=n_draw, replace=True))
    boot_idx = np.concatenate(boot_parts)
    oob_idx  = np.setdiff1d(all_idx, np.unique(boot_idx))

    print(f'\n{"─"*58}')
    print(f'  Member {m_idx+1}/{N_ENSEMBLE}   |   bootstrap={len(boot_idx):,}   OOB={len(oob_idx):,}')
    print(f'{"─"*58}')

    boot_loader = DataLoader(Subset(train_dataset, boot_idx),
                             batch_size=BATCH_SIZE, shuffle=True,
                             num_workers=NW, pin_memory=pin)
    oob_loader  = DataLoader(_DetView(train_dataset, oob_idx),
                             batch_size=BATCH_SIZE, shuffle=False,
                             num_workers=NW, pin_memory=pin)

    torch.manual_seed(SEED + 100 + m_idx)   # different init + augmentation stream per member
    model, optimizer, scheduler = make_model()
    best_acc, best_state, patience_ctr = -1.0, None, 0

    for epoch in range(NUM_EPOCHS):
        tr_loss, tr_acc = train_one_epoch(model, boot_loader, optimizer)
        oob_loss, oob_acc, _, _ = evaluate(model, oob_loader)
        scheduler.step()
        print(f'  Ep {epoch+1:02d}/{NUM_EPOCHS}  '
              f'Train Loss={tr_loss:.4f} Acc={tr_acc:.4f}  |  '
              f'OOB   Loss={oob_loss:.4f} Acc={oob_acc:.4f}')
        if oob_acc > best_acc:
            best_acc, patience_ctr = oob_acc, 0
            best_state = {k2: v.cpu() for k2, v in model.state_dict().items()}
        else:
            patience_ctr += 1
            if patience_ctr >= PATIENCE:
                print(f'  Early stopping at epoch {epoch+1}.')
                break

    bootstrap_states.append(best_state)
    oob_accs.append(best_acc)
    print(f'  → Member {m_idx+1} best OOB acc: {best_acc:.4f}')
    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

print(f'\nOOB accuracies : {[f"{a:.4f}" for a in oob_accs]}')
print(f'Mean OOB acc   : {np.mean(oob_accs):.4f} ± {np.std(oob_accs):.4f}')


# ── Aggregate: soft-voting ensemble (+ optional flip TTA) ────────────────────
fold_members    = fold_states if (INCLUDE_FOLD_MODELS and 'fold_states' in globals()) else []
ensemble_states = list(fold_members) + bootstrap_states
print(f'\nEnsemble members: {len(ensemble_states)}  '
      f'({len(fold_members)} k-fold  +  {len(bootstrap_states)} bootstrap)')


class EnsembleModel(nn.Module):
    """Bagging aggregator: averages the members' softmax probabilities
    (optionally also over horizontal-flip TTA views). forward() returns
    log-probabilities, so argmax / CrossEntropyLoss / evaluate() keep
    working exactly as for a single model."""
    def __init__(self, states, tta=True):
        super().__init__()
        self.members = nn.ModuleList()
        for st in states:
            m = RajukResNet(len(CLASS_NAMES), dropout=DROPOUT)
            m.load_state_dict({k: v.cpu() for k, v in st.items()})
            self.members.append(m)
        self.tta = tta

    def forward(self, x):
        views = [x, torch.flip(x, dims=[3])] if self.tta else [x]
        probs = torch.zeros(x.size(0), len(CLASS_NAMES), device=x.device)
        for m in self.members:
            for v in views:
                probs = probs + torch.softmax(m(v), dim=1)
        probs = probs / (len(self.members) * len(views))
        return torch.log(probs.clamp_min(1e-9))   # log-probs behave like logits


final_model = EnsembleModel(ensemble_states, tta=TTA_ENABLED).to(device).eval()

val_loss,  val_acc,  val_preds,  val_true  = evaluate(final_model, val_loader)
test_loss, test_acc, test_preds, test_true = evaluate(final_model, test_loader)

print(f'\n{"="*50}')
print(f'  BAGGED ENSEMBLE — {len(ensemble_states)} members'
      + (', flip-TTA' if TTA_ENABLED else ''))
print(f'  Validation — Loss: {val_loss:.4f}   Acc: {val_acc:.4f}')
print(f'  Test       — Loss: {test_loss:.4f}   Acc: {test_acc:.4f}')
print(f'{"="*50}')

# ── Variance-reduction check: single member vs ensemble ─────────────────────
single_model = EnsembleModel(bootstrap_states[:1], tta=False).to(device).eval()
_, s_val_acc, _, _ = evaluate(single_model, val_loader)
print(f'\nSingle bootstrap member — Val Acc: {s_val_acc:.4f}')
print(f'Ensemble gain over single model : {val_acc - s_val_acc:+.4f}  (variance removed by bagging)')
del single_model
if torch.cuda.is_available():
    torch.cuda.empty_cache()
```

## 13b. Final Error Report — Train / Validation / Test

A single consolidated report of the **final fine-tuned model** (`final_model`).
Training error is measured on the training images with the *deterministic*
(non-augmenting) transform, so it reflects the model's true fit rather than
augmentation noise.  `error = 1 − accuracy`.

```python
# ── Deterministic-transform view of the TRAIN set (reuses cached PIL images) ──
class _EvalView(Dataset):
    """Wraps an already-cached RajukDataset but applies the val/test transform
    (no random augmentation) so training error is measured cleanly."""
    def __init__(self, base_ds, transform):
        self.images, self.labels, self.transform = base_ds.images, base_ds.labels, transform
    def __len__(self):
        return len(self.labels)
    def __getitem__(self, idx):
        return self.transform(self.images[idx]), self.labels[idx]

train_eval_loader = DataLoader(
    _EvalView(train_dataset, val_test_transform),
    batch_size=BATCH_SIZE, shuffle=False, num_workers=NW, pin_memory=pin
)

tr_loss_f,  tr_acc_f,  _, _ = evaluate(final_model, train_eval_loader)
val_loss_f, val_acc_f, _, _ = evaluate(final_model, val_loader)
te_loss_f,  te_acc_f,  _, _ = evaluate(final_model, test_loader)

report_rows = [
    ('Training',   len(train_dataset), tr_loss_f,  tr_acc_f,  1 - tr_acc_f),
    ('Validation', len(val_dataset),   val_loss_f, val_acc_f, 1 - val_acc_f),
    ('Test',       len(test_dataset),  te_loss_f,  te_acc_f,  1 - te_acc_f),
]

print('=' * 68)
print('  FINAL ERROR REPORT  —  model = final_model (fine-tuned best CV weights)')
print('=' * 68)
print(f'  {"Split":<12}{"Samples":>9}{"Loss":>12}{"Accuracy":>12}{"Error":>12}')
print('  ' + '-' * 64)
for name, n, loss, acc, err in report_rows:
    print(f'  {name:<12}{n:>9,}{loss:>12.4f}{acc:>11.2%}{err:>12.2%}')
print('=' * 68)

# ── Bar chart of the three error rates ──────────────────────────────
names  = [r[0] for r in report_rows]
errors = [r[4] for r in report_rows]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

fig, ax = plt.subplots(figsize=(7, 4.5))
bars = ax.bar(names, errors, color=colors, width=0.6, edgecolor='black', linewidth=0.7)
for b, err in zip(bars, errors):
    ax.text(b.get_x() + b.get_width() / 2, b.get_height() + 0.005,
            f'{err:.2%}', ha='center', va='bottom', fontsize=11, fontweight='bold')
ax.set_ylabel('Error rate  (1 − accuracy)', fontsize=11)
ax.set_title('Final Model — Train / Validation / Test Error', fontsize=12, fontweight='bold')
ax.set_ylim(0, max(errors) * 1.25 + 0.02)
ax.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.savefig('final_error_report.png', dpi=130, bbox_inches='tight')
plt.show()
```

| 🟠 Orange triangle | **Validation error** |
| 🟢 Green square | **Test error** |


```python
def train_and_measure(n_train: int):
    actual_n = min(n_train, len(train_dataset))
    if actual_n < len(CLASS_NAMES):
        return None

    sub_idx, _ = train_test_split(
        np.arange(len(train_dataset)),
        train_size=actual_n,
        random_state=SEED,
        stratify=train_labels_arr
    )
    sub_loader = DataLoader(
        Subset(train_dataset, sub_idx),
        batch_size=BATCH_SIZE, shuffle=True, num_workers=NW, pin_memory=pin
    )

    m, opt, sch = make_model()
    CURVE_EPOCHS = min(8, NUM_EPOCHS)   # reduced from 15 → faster learning-curve sweep

    for _ in range(CURVE_EPOCHS):
        train_one_epoch(m, sub_loader, opt)
        sch.step()

    tr_loss, _, _, _ = evaluate(m, sub_loader)
    vl_loss, _, _, _ = evaluate(m, val_loader)
    te_loss, _, _, _ = evaluate(m, test_loader)
    return tr_loss, vl_loss, te_loss


curve_sizes, curve_train, curve_val, curve_test = [], [], [], []

for n in DATASET_STEPS:
    if n > len(train_dataset):
        print(f'  n={n}: exceeds train size ({len(train_dataset)}), skipping.')
        break
    print(f'  Training on n={n} …', end=' ', flush=True)
    result = train_and_measure(n)
    if result is None:
        print('skipped (too few samples).'); continue
    tr_l, vl_l, te_l = result
    curve_sizes.append(n)
    curve_train.append(tr_l)
    curve_val.append(vl_l)
    curve_test.append(te_l)
    print(f'train={tr_l:.4f}  val={vl_l:.4f}  test={te_l:.4f}')

print('\nLearning-curve data collected.')
```

```python
fig, ax = plt.subplots(figsize=(11, 6))

ax.plot(curve_sizes, curve_train, color='#1f77b4', linewidth=1.8, linestyle='--', alpha=0.6)
ax.plot(curve_sizes, curve_val,   color='#ff7f0e', linewidth=1.8, linestyle='--', alpha=0.6)
ax.plot(curve_sizes, curve_test,  color='#2ca02c', linewidth=1.8, linestyle='--', alpha=0.6)

ax.scatter(curve_sizes, curve_train, marker='o', s=100,
           color='#1f77b4', zorder=5, label='Training Error')
ax.scatter(curve_sizes, curve_val,   marker='^', s=110,
           color='#ff7f0e', zorder=5, label='Validation Error')
ax.scatter(curve_sizes, curve_test,  marker='s', s=100,
           color='#2ca02c', zorder=5, label='Test Error')

for s in curve_sizes:
    ax.axvline(s, color='grey', linewidth=0.4, linestyle=':', alpha=0.5)

ax.set_xlabel('Number of Training Samples', fontsize=12)
ax.set_ylabel('Cross-Entropy Loss', fontsize=12)
ax.set_title('Learning Curve — Train / Validation / Test Error\n'
             '(3 dots per dataset size: ● Train  ▲ Val  ■ Test)',
             fontsize=13, fontweight='bold')
ax.legend(fontsize=11, framealpha=0.9)
ax.set_xticks(curve_sizes)
ax.tick_params(axis='x', rotation=30)
ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('learning_curve.png', dpi=130, bbox_inches='tight')
plt.show()
print('Learning curve saved.')
```

## 14b. Bias–Variance Analysis

Using the learning-curve measurements above we visualise the classic
**bias–variance decomposition** of the model's error:

* **Bias**  →  the error floor the *training* curve settles at (how well the model
  can fit the data at all — a high floor ⇒ under-fitting).
* **Variance**  →  the **gap** between the validation and training error
  (how much performance drops on unseen data — a large gap ⇒ over-fitting).

The shaded band is the variance gap; the dashed line marks the bias floor. The right
panel stacks the two components to show how they add up to the total validation error.

```python
# ── Bias–Variance visualisation (derived from the learning-curve sweep) ────
if len(curve_sizes) < 1:
    print('No learning-curve data — run the learning-curve cells (section 14) first.')
else:
    sizes = np.asarray(curve_sizes, dtype=float)
    tr    = np.asarray(curve_train, dtype=float)
    vl    = np.asarray(curve_val,   dtype=float)

    bias_floor   = float(tr[-1])                    # train error at largest size ≈ bias
    variance_gap = float(vl[-1] - tr[-1])           # val − train gap    ≈ variance
    var_disp     = max(variance_gap, 0.0)           # clamp for the stacked bar

    fig, (axL, axR) = plt.subplots(1, 2, figsize=(14, 5.2),
                                   gridspec_kw={'width_ratios': [1.6, 1]})

    # Left: train vs val error, variance gap shaded, bias floor marked
    axL.plot(sizes, tr, 'o-', color='#1f77b4', linewidth=2, label='Training error (→ bias)')
    axL.plot(sizes, vl, '^-', color='#ff7f0e', linewidth=2, label='Validation error')
    axL.fill_between(sizes, tr, vl, color='#ff7f0e', alpha=0.15,
                     label='Variance (val − train gap)')
    axL.axhline(bias_floor, color='#1f77b4', linestyle='--', linewidth=1.2, alpha=0.8)
    axL.annotate('Bias floor', xy=(sizes[0], bias_floor),
                 xytext=(sizes[0], bias_floor + 0.03),
                 fontsize=10, color='#1f77b4', fontweight='bold')
    axL.set_xlabel('Number of Training Samples', fontsize=11)
    axL.set_ylabel('Cross-Entropy Loss (error)', fontsize=11)
    axL.set_title('Bias–Variance: Train vs Validation Error', fontsize=12, fontweight='bold')
    axL.set_xticks(sizes)
    axL.tick_params(axis='x', rotation=30)
    axL.legend(fontsize=10, framealpha=0.9)
    axL.grid(axis='y', linestyle='--', alpha=0.4)

    # Right: decomposition of the final error into bias + variance
    axR.bar(0, bias_floor, color='#1f77b4', edgecolor='black', linewidth=0.7, width=0.55)
    axR.bar(1, var_disp,   color='#ff7f0e', edgecolor='black', linewidth=0.7, width=0.55)
    axR.bar(2, bias_floor, color='#1f77b4', edgecolor='black', linewidth=0.7, width=0.55)
    axR.bar(2, var_disp, bottom=bias_floor, color='#ff7f0e',
            edgecolor='black', linewidth=0.7, width=0.55)
    for x, val in [(0, bias_floor), (1, var_disp), (2, bias_floor + var_disp)]:
        axR.text(x, val + 0.01, f'{val:.3f}', ha='center', va='bottom',
                 fontsize=10, fontweight='bold')
    axR.set_xticks([0, 1, 2])
    axR.set_xticklabels(['Bias\n(train err)', 'Variance\n(gap)', 'Total\n(val err)'])
    axR.set_ylabel('Cross-Entropy Loss', fontsize=11)
    axR.set_title('Error Decomposition (largest train size)', fontsize=12, fontweight='bold')
    axR.grid(axis='y', linestyle='--', alpha=0.4)

    plt.tight_layout()
    plt.savefig('bias_variance.png', dpi=130, bbox_inches='tight')
    plt.show()

    print(f'Bias  (training-error floor) : {bias_floor:.4f}')
    print(f'Variance (val − train gap)    : {variance_gap:.4f}')
    print(f'Total validation error       : {float(vl[-1]):.4f}')
```

## 15. Full Training History (Final Model)

```python
history_model, history_opt, history_sch = make_model()
h = {'train_loss': [], 'val_loss': [], 'test_loss': [],
     'train_acc':  [], 'val_acc':  [], 'test_acc':  []}

best_acc_ht   = -1.0
best_state_ht = None
patience_ht   = 0

print(f'Training for up to {NUM_EPOCHS} epochs …\n')
for epoch in range(NUM_EPOCHS):
    tr_l, tr_a = train_one_epoch(history_model, train_loader, history_opt)
    vl_l, vl_a, _, _ = evaluate(history_model, val_loader)
    te_l, te_a, _, _ = evaluate(history_model, test_loader)
    history_sch.step()

    for key, val in zip(
        ['train_loss','val_loss','test_loss','train_acc','val_acc','test_acc'],
        [tr_l, vl_l, te_l, tr_a, vl_a, te_a]
    ):
        h[key].append(val)

    print(f'Ep {epoch+1:02d}/{NUM_EPOCHS}  '
          f'Train L={tr_l:.4f} A={tr_a:.4f} | '
          f'Val L={vl_l:.4f} A={vl_a:.4f} | '
          f'Test L={te_l:.4f} A={te_a:.4f}')

    if vl_a > best_acc_ht:
        best_acc_ht   = vl_a
        patience_ht   = 0
        best_state_ht = {k2: v.cpu() for k2, v in history_model.state_dict().items()}
    else:
        patience_ht += 1
        if patience_ht >= PATIENCE:
            print(f'Early stopping at epoch {epoch+1}.')
            break

print(f'\nBest validation accuracy: {best_acc_ht:.4f}')
```

```python
epochs_ran = range(1, len(h['train_loss']) + 1)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(epochs_ran, h['train_loss'], 'o-', color='#1f77b4', label='Train Loss')
ax1.plot(epochs_ran, h['val_loss'],   '^-', color='#ff7f0e', label='Val Loss')
ax1.plot(epochs_ran, h['test_loss'],  's-', color='#2ca02c', label='Test Loss')
ax1.set_title('Loss per Epoch  (Train / Val / Test)', fontsize=12, fontweight='bold')
ax1.set_xlabel('Epoch'); ax1.set_ylabel('Cross-Entropy Loss')
ax1.legend(); ax1.grid(linestyle='--', alpha=0.4)

ax2.plot(epochs_ran, h['train_acc'], 'o-', color='#1f77b4', label='Train Acc')
ax2.plot(epochs_ran, h['val_acc'],   '^-', color='#ff7f0e', label='Val Acc')
ax2.plot(epochs_ran, h['test_acc'],  's-', color='#2ca02c', label='Test Acc')
ax2.set_title('Accuracy per Epoch  (Train / Val / Test)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Epoch'); ax2.set_ylabel('Accuracy')
ax2.legend(); ax2.grid(linestyle='--', alpha=0.4)

plt.tight_layout()
plt.savefig('training_history.png', dpi=130, bbox_inches='tight')
plt.show()
```

## 17. Sanity Check — Predict on Random Test Samples

```python
best_model = RajukResNet(len(CLASS_NAMES), dropout=DROPOUT).to(device)
best_model.load_state_dict({k: v.to(device) for k, v in best_state_ht.items()})
best_model.eval()

rng2   = np.random.default_rng(SEED + 1)
sample = rng2.choice(len(test_dataset), size=min(5, len(test_dataset)), replace=False)

fig, axes = plt.subplots(1, len(sample), figsize=(len(sample) * 3.5, 4))
fig.suptitle('Sanity Check — Random Test Predictions', fontsize=13, fontweight='bold')

# Un-normalise for display
inv_mean = torch.tensor(IMAGENET_MEAN).view(3,1,1)
inv_std  = torch.tensor(IMAGENET_STD).view(3,1,1)

with torch.no_grad():
    for ax, s_idx in zip(axes, sample):
        img_tensor, true_lbl = test_dataset[int(s_idx)]
        logits   = best_model(img_tensor.unsqueeze(0).to(device))
        pred_lbl = logits.argmax(1).item()

        # Reverse ImageNet normalisation for display
        img_display = (img_tensor * inv_std + inv_mean).clamp(0, 1)
        img_np = img_display.permute(1, 2, 0).numpy()

        ax.imshow(img_np)
        ax.set_xticks([]); ax.set_yticks([])

        colour = '#2ca02c' if pred_lbl == true_lbl else '#d62728'
        ax.set_title(
            f'True:  {CLASS_NAMES[true_lbl]}\nPred:  {CLASS_NAMES[pred_lbl]}',
            fontsize=9, color=colour, fontweight='bold'
        )
        for spine in ax.spines.values():
            spine.set_edgecolor(colour); spine.set_linewidth(3)

plt.tight_layout()
plt.savefig('sanity_predictions.png', dpi=120, bbox_inches='tight')
plt.show()
print('Sanity check complete.')
```
